## Prepare data for modeling

### Subtask:
Split the data into features (X) and target (y), and handle any necessary data transformations or scaling.


**Reasoning**:
Drop the 'Country' column, handle missing values by dropping rows with NaN, define features (X) and target (y), and then scale the features.



```python
# Drop the 'Country' column
df = df.drop('Country', axis=1)

# Handle missing values by dropping rows with any NaN values
df = df.dropna()

# Define the target variable y
y = df['2025']

# Define the features X
X = df.drop('2025', axis=1)

# Import and instantiate StandardScaler
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

# Fit and transform the features
X_scaled = scaler.fit_transform(X)

display(X.head())
display(y.head())
display(X_scaled[:5])
```

## Train and test models

### Subtask:
Train several machine learning models on the data and evaluate their performance.


**Reasoning**:
Import necessary libraries, split the data into training and testing sets, train three different regression models, make predictions, and evaluate their performance.



```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, r2_score

# Split the data
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Instantiate and train models
models = {
    "Linear Regression": LinearRegression(),
    "Random Forest Regressor": RandomForestRegressor(random_state=42),
    "Gradient Boosting Regressor": GradientBoostingRegressor(random_state=42)
}

model_performance = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    model_performance[name] = {"MSE": mse, "R2": r2}

# Print model performance
print("Model Performance:")
for name, metrics in model_performance.items():
    print(f"{name}:")
    print(f"  MSE: {metrics['MSE']:.2f}")
    print(f"  R2: {metrics['R2']:.2f}")
    print("-" * 20)
```

## Evaluate model performance

### Subtask:
Assess the precision and other relevant metrics for each model.


**Reasoning**:
Assess the performance of each model based on the calculated MSE and R2 values.



```python
# Analyze the Mean Squared Error (MSE) and R-squared (R2) values for each model

print("Model Performance Analysis:")
print("-" * 30)

for name, metrics in model_performance.items():
    print(f"{name}:")
    print(f"  MSE: {metrics['MSE']:.2f}")
    print(f"  R2: {metrics['R2']:.2f}")
    print("-" * 20)

# Compare the performance and identify the best model
print("\nModel Comparison:")
print("-" * 30)

best_model_r2 = max(model_performance, key=lambda k: model_performance[k]['R2'])
best_model_mse = min(model_performance, key=lambda k: model_performance[k]['MSE'])

print(f"Model with the highest R2 score: {best_model_r2} ({model_performance[best_model_r2]['R2']:.2f})")
print(f"Model with the lowest MSE: {best_model_mse} ({model_performance[best_model_mse]['MSE']:.2f})")

print("\nImportance of MSE and R2:")
print("-" * 30)
print("Mean Squared Error (MSE) measures the average squared difference between the actual and predicted values. A lower MSE indicates better model performance in terms of prediction accuracy.")
print("R-squared (R2) represents the proportion of the variance in the dependent variable that is predictable from the independent variables. An R2 score closer to 1 indicates that the model explains a larger portion of the variance, suggesting a better fit to the data.")

# Summarize the findings on model performance
print("\nSummary of Model Performance Findings:")
print("-" * 30)
print(f"Based on the evaluation metrics:")
print(f"- {best_model_r2} achieved the highest R2 score ({model_performance[best_model_r2]['R2']:.2f}), indicating it explains the most variance in the target variable.")
print(f"- {best_model_mse} achieved the lowest MSE ({model_performance[best_model_mse]['MSE']:.2f}), indicating the smallest average squared prediction error.")
print("All models demonstrated strong performance with high R2 scores, suggesting they are good fits for the data. The Linear Regression model shows a remarkably low MSE, which could be due to the nature of the data and features used.")
```

```python
# 1. Identify the best performing model
# Based on the previous evaluation, Linear Regression had the lowest MSE and highest R2.
best_model_name = min(model_performance, key=lambda k: model_performance[k]['MSE'])
best_model = models[best_model_name]

print(f"The best performing model based on MSE and R2 is: {best_model_name}")

# 2. Use the best performing model to make predictions on the test set (X_test is already scaled)
y_pred = best_model.predict(X_test)

# 3. Create a DataFrame from the test set, including the actual target values (y_test) and the predicted values.
# Need to get the original indices from y_test to merge later.
results_df = pd.DataFrame({'Actual': y_test, 'Predicted': y_pred}, index=y_test.index)
results_df['Absolute_Error'] = abs(results_df['Actual'] - results_df['Predicted'])

# 4. Merge this DataFrame with the original DataFrame df (before dropping the 'Country' column and handling NaNs) based on the index to associate predictions with countries.
# Need to reload the original df before dropping 'Country' and NaNs for merging.
original_df = pd.read_csv('/content/2020-2025.csv')

# Merge results_df with original_df to get country names.
# We need to merge based on the index.
results_with_country = original_df.loc[results_df.index].copy()
results_with_country['Actual_2025'] = results_df['Actual']
results_with_country['Predicted_2025'] = results_df['Predicted']
results_with_country['Absolute_Error_2025'] = results_df['Absolute_Error']


# 6. Analyze the errors to identify which countries have the lowest prediction errors and highest errors.
# Sort by absolute error to find best and worst predictions
best_predicted_countries = results_with_country.nsmallest(5, 'Absolute_Error_2025')
worst_predicted_countries = results_with_country.nlargest(5, 'Absolute_Error_2025')

print("\nCountries with the lowest prediction errors (Best Predicted):")
display(best_predicted_countries[['Country', 'Actual_2025', 'Predicted_2025', 'Absolute_Error_2025']])

print("\nCountries with the highest prediction errors (Worst Predicted):")
display(worst_predicted_countries[['Country', 'Actual_2025', 'Predicted_2025', 'Absolute_Error_2025']])

# 7. Summarize the findings
print("\nSummary of Prediction Analysis:")
print("-" * 30)
print(f"The {best_model_name} model was used to predict the 2025 GDP values.")
print("Countries with the lowest absolute prediction errors are considered best predicted.")
print("Countries with the highest absolute prediction errors may exhibit interesting patterns or outliers that the model struggled to capture.")
print(f"\nBased on the analysis, the countries with the lowest prediction errors are:")
for index, row in best_predicted_countries.iterrows():
    print(f"- {row['Country']}: Absolute Error = {row['Absolute_Error_2025']:,.2f}")

print(f"\nConversely, the countries with the highest prediction errors are:")
for index, row in worst_predicted_countries.iterrows():
    print(f"- {row['Country']}: Absolute Error = {row['Absolute_Error_2025']:,.2f}")
```
